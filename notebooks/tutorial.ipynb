## D.1 From Oscillator to Phase Variable

$x(t) = A(t) \cos(2\pi\nu_0 t + \phi(t))$

### Exercise D.1.1

Q - quality factor (Frequency Standards, p.97):

$$
Q = \omega \frac{W}{-dW/dt}
$$

Differential equation: 

$$
\frac{1}{W}dW = -\frac {\omega_0}{Q} dt
$$

$$
\ln(W) = -\frac {\omega_0}{Q} t + C
$$

$$
W(t) = W_0 e^{-\frac {\omega_0}{Q} t}
$$

Energy of the harmonic oscillator: 

$$
W = k A^2(t)
$$

Insert this into solution of differential equation:

$$
k A^2(t) = kA^2_0 e^{-\frac {\omega_0}{Q} t}
$$

$$
A(t) = A_0 e^{-\frac {\omega_0}{2Q} t}
$$

Amplitude fluctuation: 

$$
\dot{A} = -\frac {\omega_0}{2Q}A_0 e^{-\frac {\omega_0}{2Q} t}
$$

$$
\dot{A} = -\frac{\omega_0}{2Q} A(t)
$$

=> Amplitude fluctuation depend on the current value of amplitude

Phase $\Phi = 2\pi\nu_0 t + \phi(t)$

where $2\pi\nu_0 t $ part corresponds to nominal change (for ideal case) and $\phi(t)$ to noise.

Phase fluctuation: $\dot{\Phi} = 2\pi\nu_0 + \dot{\phi(t)}$

=> the nominal part is constant and the noise part does not depend on $\phi(t)$ unlike the amplitude.

Sources: 
Fritz Riehle, Frequency Standards: Basics and Applications, Wiley-VCH, 2003.
